```python
a=10
print(a)
print(type(a))
```

```python
#type conversion or type casting
#implicite type conversyion
#automatic type conversion-->low to highdata types
a=10
print(a,type(a))
a=a/3
print(a,type(a))
```

```python
#explict type conversion
#manual type conversion->higher to lower data types
a=10.5  #float-->8 bytes
print(a,type(a))
a= int(a)# int--->4 bytes
#manual conversion using int() function
print(a,type(a))
```
